```python
from langchain_qdrant import QdrantVectorStore
from qdrant_client import QdrantClient
```

```python
qdrant = QdrantVectorStore.from_existing_collection("db/test_rag")
```

```python
from collections import Counter

name = "test_rag"
path = f"db/{name}"
client = QdrantClient(path=path)

points, _ = client.scroll(
    collection_name=name,
    limit=1000,
    with_payload=True,
    with_vectors=False,
)

by_type = Counter()
by_source = Counter()

for pt in points:
    payload = pt.payload if hasattr(pt, "payload") else pt.get("payload", {})
    if not payload:
        continue
    by_type[payload.get("type")] += 1
    by_source[payload.get("source")] += 1

print("by_type:", by_type)
print("by_source (top):", by_source.most_common(10))
```
